#### Cross data consistency
Order ID : for the same order we can not find different Order IDs which is true in our Data Set

Customer ID Customer Name : we must have the same number of Customer ID and Customer Name wxhich is true 

Product ID and Product Name :  we must have the same number of Product ID and   Product Name which is False  we have more ditinct Product ID than Distinc Product Name which means there are some product which have different IDs


```python
ColumnsToCheck2=SuperStore.columns.drop(['Row ID' , 'Order ID'])
print("The number of duplicated orders is " , SuperStore.duplicated(subset=ColumnsToCheck2, keep=False).sum())
print(SuperStore["Customer ID"].nunique() == SuperStore["Customer Name"].nunique() )
print(SuperStore["Product ID"].nunique(),SuperStore["Product Name"].nunique() )
CountD=SuperStore.groupby("Product Name")["Product ID"].nunique().reset_index()
NamesIDs=CountD[CountD["Product ID"] > 1]
print(NamesIDs)
```

as we can see here we have 60 row which have a probleme in their product id we will standrize them for every row with a product name x we will assigne for it the same product id y 

```python
ProblemNames= NamesIDs["Product Name"]
TargetMap = SuperStore[SuperStore['Product Name'].isin(ProblemNames)].groupby("Product Name")["Product ID"].first().to_dict()
SuperStore.loc[SuperStore["Product Name"].isin(ProblemNames) , "Product ID"]=SuperStore["Product Name"].map(TargetMap)
SuperStore["Product ID"].nunique()-SuperStore["Product Name"].nunique()
```

as we can see the result is -32 that means that we have a probleme in the other direction one product id has multiple names attached to it 


```python
countN=SuperStore.groupby("Product ID")["Product Name"].nunique().reset_index()
print(countN)
IDsNames=countN[countN["Product Name"]>1]
print(IDsNames.sum())
```

```python
SuperStore=SuperStore.copy()
ProblemIDs= IDsNames["Product ID"]
TargetMap = SuperStore[SuperStore['Product ID'].isin(ProblemIDs)].groupby("Product ID")["Product Name"].first().to_dict()
SuperStore.loc[SuperStore["Product ID"].isin(ProblemIDs) , "Product Name"]=SuperStore["Product ID"].map(TargetMap)
SuperStore["Product ID"].nunique()-SuperStore["Product Name"].nunique()
```
