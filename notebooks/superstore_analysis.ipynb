### Time_series analysis
we have two dates we will consider the order dat not shipping date because the paiement is due wehen ordering

```python
MonthlySales=SuperStore.groupby(SuperStore["Order Date"].dt.to_period("M") )["Sales"].sum().reset_index()
print("The monthly sales are :\n" , MonthlySales)
MonthlyProfit=SuperStore.groupby(SuperStore["Order Date"].dt.to_period("M"))["Profit"].sum().reset_index()
print("The monthly sales are :\n " , MonthlyProfit)
print("the month with the most sales is :" , MonthlySales.sort_values( by="Sales",ascending=False).values[0])
print("the month with the least sales is :" , MonthlySales.sort_values( by="Sales").values[0])
yearlySales=SuperStore.groupby(SuperStore["Order Date"].dt.year)["Sales"].sum().reset_index()
print("The yearly sales are :\n" , yearlySales)
yearlyProfit=SuperStore.groupby(SuperStore["Order Date"].dt.year)["Profit"].sum().reset_index()
print("The yearly Profits are :\n" , yearlyProfit)
```

As we can see the sales downgrade in the second year but incremants by plus than 25% in the years that follows although the profit is in an increasing trend so we can conclude that in the first half the do not follow the same pattern but in the second half they do 
by this stats we can conclude that less sales does not mean necessarly less profit as we can see in the ratio table there is 2015 which has less sales but bigger profit than 2014 which means the most important factor of the bussines in our data set is not sales but it is profit

```python
Ratio=yearlySales.merge(yearlyProfit , how ="left" , on = 'Order Date')
Ratio['Ratio']=(Ratio["Profit"]/Ratio["Sales"])*100
Ratio.sort_values(by="Ratio")
```
